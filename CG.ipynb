#Prb 01: DDA Algorithm

```python
import matplotlib.pyplot as plt

def dda(x1, y1, x2, y2):

    dx = x2 - x1
    dy = y2 - y1

    steps = max(abs(dx), abs(dy))

    x_inc = dx / steps
    y_inc = dy / steps

    x, y = x1, y1

    x_points = []
    y_points = []

    for i in range(steps + 1):

        x_points.append(round(x + 0.0000001))
        y_points.append(round(y + 0.0000001))

        x = x + x_inc
        y = y + y_inc

    plt.plot(x_points, y_points, 'ro-')
    print((x_points,y_points))
    plt.grid(True)
    plt.show()


dda(5, 6, 13, 10)
```

#Prb 02 Bresenham Line Draw

```python
import matplotlib.pyplot as plt

def bresenham(x1, y1, x2, y2):

    dx = x2 - x1
    dy = y2 - y1

    p = 2 * dy - dx

    x,y = x1,y1


    x_points = []
    y_points = []

    for i in range(dx + 1):

        x_points.append(x)
        y_points.append(y)

        if p < 0:
            p = p + 2 * dy
            x = x + 1
            y = y

        else:
            p = p + 2 * dy - 2 * dx
            x = x + 1
            y = y + 1

    print(x_points, y_points)

    plt.plot(x_points, y_points, 'ro-')
    plt.grid(True)
    plt.show()


bresenham(9, 18, 14, 22)
```

#Prb 03: Bresenham Circle Drawing Algorithm

```python
import matplotlib.pyplot as plt

def midpoint_circle(xc, yc, r):

    x = 0
    y = r

    p = 1 - r

    x_points = []
    y_points = []

    while x <= y:

        points = [
            (xc + x, yc + y),
            (xc - x, yc + y),
            (xc + x, yc - y),
            (xc - x, yc - y),
            (xc + y, yc + x),
            (xc - y, yc + x),
            (xc + y, yc - x),
            (xc - y, yc - x)
        ]

        for px, py in points:
            x_points.append(px)
            y_points.append(py)

        if p < 0:
            x = x + 1
            p = p + 2 * x + 1

        else:
            x = x + 1
            y = y - 1
            p = p + 2 * x - 2 * y + 1

    plt.scatter(x_points, y_points)
    plt.grid(True)
    plt.axis('equal')
    plt.title("Midpoint Circle Drawing Algorithm")
    plt.show()


midpoint_circle(0, 0, 10)
```

#Prb 04:  Bresenha circle draw algorithm

```python

import matplotlib.pyplot as plt

def bresenham_circle(xc, yc, r):

    x = 0
    y = r

    p = 3 - 2 * r

    x_points = []
    y_points = []

    while x <= y:

        points = [
            (xc + x, yc + y),
            (xc - x, yc + y),
            (xc + x, yc - y),
            (xc - x, yc - y),
            (xc + y, yc + x),
            (xc - y, yc + x),
            (xc + y, yc - x),
            (xc - y, yc - x)
        ]

        for px, py in points:
            if (px, py) not in zip(x_points, y_points):
                x_points.append(px)
                y_points.append(py)

        if p < 0:
            x = x + 1
            p = p + 4 * x + 6

        else:
            x = x + 1
            y = y - 1
            p = p + 4 * (x - y) + 10

    print("X Points:", x_points)
    print("Y Points:", y_points)

    plt.scatter(x_points, y_points, color='red')
    plt.grid(True)
    plt.axis('equal')
    plt.title("Bresenham Circle Drawing Algorithm")
    plt.xlabel("X-axis")
    plt.ylabel("Y-axis")
    plt.show()


bresenham_circle(0, 0, 8)
```

#Rotation

```python

import matplotlib.pyplot as plt
import math

# Original triangle points
x = [1, 4, 2, 1]
y = [1, 1, 3, 1]

# Take any angle
angle = 90

# Convert degree to radian
r = math.radians(angle)

# Rotation formula
x1 = []
y1 = []

for i in range(len(x)):
    new_x = x[i]*math.cos(r) - y[i]*math.sin(r)
    new_y = x[i]*math.sin(r) + y[i]*math.cos(r)

    x1.append(new_x)
    y1.append(new_y)

# Draw original triangle
plt.plot(x, y, 'bo-', label='Original')

# Draw rotated triangle
plt.plot(x1, y1, 'ro-', label='Rotated')

plt.grid()
plt.axhline(0)
plt.axvline(0)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Rotation Transformation")

plt.legend()
plt.axis('equal')
plt.show()
```

#Scaling

```python

import matplotlib.pyplot as plt

# Original triangle points
x = [1, 4, 2, 1]
y = [1, 1, 3, 1]

# Scaling factors
sx = 2
sy = 2

# Scaling transformation
x1 = []
y1 = []

for i in range(len(x)):
    new_x = x[i] * sx
    new_y = y[i] * sy

    x1.append(new_x)
    y1.append(new_y)

# Original triangle
plt.plot(x, y, 'bo-', label='Original')

# Scaled triangle
plt.plot(x1, y1, 'ro-', label='Scaled')

plt.grid()
plt.axhline(0)
plt.axvline(0)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Scaling Transformation")

plt.legend()
plt.axis('equal')
plt.show()
```

#Reflection

```python

import matplotlib.pyplot as plt

# Original triangle points
x = [1, 4, 2, 1]
y = [1, 1, 3, 1]

# Choose reflection type
choice = 1

x1 = []
y1 = []

for i in range(len(x)):

    if choice == 1:       # X-axis
        new_x = x[i]
        new_y = -y[i]

    elif choice == 2:     # Y-axis
        new_x = -x[i]
        new_y = y[i]

    elif choice == 3:     # Origin
        new_x = -x[i]
        new_y = -y[i]

    elif choice == 4:     # Line y = x
        new_x = y[i]
        new_y = x[i]

    x1.append(new_x)
    y1.append(new_y)

# Draw original triangle
plt.plot(x, y, 'bo-', label='Original')

# Draw reflected triangle
plt.plot(x1, y1, 'ro-', label='Reflected')

plt.grid()
plt.axhline(0, color='black')
plt.axvline(0, color='black')

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("2D Reflection Transformation")

plt.legend()
plt.axis('equal')
plt.show()
```

```python

import matplotlib.pyplot as plt


def inside(point, boundary, value):

    x, y = point

    if boundary == 'LEFT':
        return x >= value

    elif boundary == 'RIGHT':
        return x <= value

    elif boundary == 'BOTTOM':
        return y >= value

    elif boundary == 'TOP':
        return y <= value


def intersection(s, p, boundary, value):

    x1, y1 = s
    x2, y2 = p

    if boundary in ['LEFT', 'RIGHT']:

        x = value
        y = y1 + (y2 - y1) * (value - x1) / (x2 - x1)

    else:

        y = value
        x = x1 + (x2 - x1) * (value - y1) / (y2 - y1)

    return (x, y)


def sutherland_hodgman(polygon, xmin, ymin, xmax, ymax):

    boundaries = [
        ('LEFT', xmin),
        ('RIGHT', xmax),
        ('BOTTOM', ymin),
        ('TOP', ymax)
    ]

    output = polygon

    for boundary, value in boundaries:

        input_list = output
        output = []

        if not input_list:
            break

        s = input_list[-1]

        for p in input_list:

            if inside(p, boundary, value):

                if inside(s, boundary, value):
                    output.append(p)

                else:
                    i = intersection(s, p, boundary, value)
                    output.append(i)
                    output.append(p)

            else:

                if inside(s, boundary, value):
                    i = intersection(s, p, boundary, value)
                    output.append(i)

            s = p

    return output


# Original polygon
polygon = [
    (2, 2),
    (8, 1),
    (10, 6),
    (6, 9),
    (1, 7)
]

# Clipping window
xmin, ymin = 3, 3
xmax, ymax = 8, 7

# Clipping operation
clipped = sutherland_hodgman(
    polygon, xmin, ymin, xmax, ymax
)

print("Clipped Polygon:", clipped)

# Plot original polygon
ox = [p[0] for p in polygon + [polygon[0]]]
oy = [p[1] for p in polygon + [polygon[0]]]

plt.plot(ox, oy, 'r--', label='Original Polygon')

# Plot clipping window
wx = [xmin, xmax, xmax, xmin, xmin]
wy = [ymin, ymin, ymax, ymax, ymin]

plt.plot(wx, wy, 'b-', label='Clipping Window')

# Plot clipped polygon
if clipped:
    cx = [p[0] for p in clipped + [clipped[0]]]
    cy = [p[1] for p in clipped + [clipped[0]]]

    plt.fill(cx, cy, color='green', alpha=0.3)
    plt.plot(cx, cy, 'g-', linewidth=2,
             label='Clipped Polygon')

plt.grid(True)
plt.axis('equal')
plt.legend()
plt.title("Sutherland-Hodgman Polygon Clipping")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.show()
```
